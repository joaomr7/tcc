```python
from pathlib import Path
import sys

# Add src to import path
PROJECT_ROOT = Path.cwd().resolve().parent
SRC_PATH = PROJECT_ROOT / 'src'
sys.path.insert(0, str(SRC_PATH))
```

```python
from langchain_openai import ChatOpenAI, OpenAIEmbeddings
from rst_kg import RKG
```

```python
# load LLM models
llm = ChatOpenAI(
    base_url='http://127.0.0.1:1234/v1',
    model='google/gemma-4-e2b',
    api_key='none',
    temperature=0,
    extra_body={
        'reasoning' : {
            'effort' : 'low'
        }
    }
)

embed = OpenAIEmbeddings(
    base_url='http://127.0.0.1:1234/v1',
    model='text-embedding-bge-m3',
    api_key='none',
    check_embedding_ctx_length=False
)
```

```python
rkg = RKG(
    llm=llm,
    embed=embed,
    dmrst_checkpoint_path='../artifacts/dmrst/multi_all_checkpoint.torchsave',
    dmrst_cache_dir='../artifacts/models'
)
```

```python
document = 'Google released Gemini 2.5 to improve reasoning and multimodal capabilities across its AI products. Several months later, Google integrated Gemini 2.5 into Google Workspace, allowing users to generate documents, summarize emails, and analyze spreadsheets.'
```

```python
final_graph = rkg.get_rkg_from_document(document)
```
